```python

from pyspark.sql import functions as F

# Customer dimension
customers = [
    (101, "Arun", "Bangalore", "South"),
    (102, "Priya", "Chennai", "South"),
    (103, "Rahul", "Mumbai", "West"),
    (104, "Sneha", "Delhi", "North"),
    (105, "Kiran", "Hyderabad", "South")
]

customer_df = spark.createDataFrame(
    customers,
    ["CustomerID", "CustomerName", "City", "Region"]
)

# Product dimension
products = [
    (201, "Laptop", "Electronics", 60000),
    (202, "Mobile", "Electronics", 25000),
    (203, "Chair", "Furniture", 8000),
    (204, "Desk", "Furniture", 12000),
    (205, "Headphones", "Accessories", 3000)
]

product_df = spark.createDataFrame(
    products,
    ["ProductID", "ProductName", "Category", "UnitPrice"]
)

# Sales fact
sales = [
    (1001, 101, 201, "2026-11-01", 1, 60000),
    (1002, 102, 202, "2026-11-02", 2, 50000),
    (1003, 103, 203, "2026-11-03", 3, 24000),
    (1004, 104, 204, "2026-11-04", 2, 24000),
    (1005, 105, 205, "2026-11-05", 4, 12000),
    (1006, 101, 202, "2026-11-06", 1, 25000),
    (1007, 103, 201, "2026-11-07", 2, 120000),
    (1008, 102, 205, "2026-11-08", 2, 6000),
    (1009, 104, 203, "2026-11-09", 1, 8000),
    (1010, 105, 204, "2026-11-10", 3, 36000)
]

sales_df = spark.createDataFrame(
    sales,
    ["OrderID", "CustomerID", "ProductID",
     "OrderDate", "Quantity", "SalesAmount"]
).withColumn(
    "OrderDate", F.to_date("OrderDate")
)

# Write Delta tables
customer_df.write.format("delta").mode("overwrite").saveAsTable("DimCustomer")

product_df.write.format("delta").mode("overwrite").saveAsTable("DimProduct")

sales_df.write.format("delta").mode("overwrite").saveAsTable("FactSales")

print("All Delta tables created successfully")

```
